````python
docs_raw=Document(page_content=readme_text,
               metadata={"repo":repo.full_name,
                         "source":"GitHub",
                         "doc_type": "raw_readme"})
````

````python
docs_analysis=Document(page_content=output,
              metadata={"repo":repo.full_name,
                         "source":"Model Analysis",
                         "doc_type": "Summary"})
````

````python
docs_raw.page_content[:100]
````

````python
docs_analysis.page_content[:100]
````

````python
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=1000,chunk_overlap=200
)
````

````python
split_docs=text_splitter.split_documents([docs_raw,docs_analysis])
````

````python
print(len(split_docs))
````

````python
embedding=GoogleGenerativeAIEmbeddings(model='gemini-embedding-001',google_api_key=api)
````

````python
vectorstore=FAISS.from_documents(split_docs,embedding)
````

````python
retriever=vectorstore.as_retriever(search_kwargs={"k": 4})
````

````python
def question(query):
    
    retrived_chunks=retriever.invoke(query)
    context="\n\n".join(chunks.page_content for chunks in retrived_chunks)
    src=set()
    repo=set()
    doc_type=set()
    
    for i in retrived_chunks:
        src.add(i.metadata['source'])
        repo.add(i.metadata['repo'])
        doc_type.add(i.metadata['doc_type'])

    meta_info={'Repositary':[repo],'Source':[src],"Document Type":[doc_type]}
    # src.add(context.metadata)
    prompt=f"""You are a repository assistant and technical explainer.
Use the retrieved context (executive architecture analysis + raw README) to answer the user’s question with accuracy and clarity.

### RULES:
1. Base answers ONLY on the provided context. No hallucination.
2. If missing, state: "This detail is not documented in the repository analysis or README."
3. Provide concise bullet points for direct answers.
4. Expand with clear teaching notes for intermediate learners:
   - Use step-by-step flows, arrows, or numbered steps for processes.
   - Use tables for comparisons or structured data.
   - Keep explanations short, accurate, and approachable.
5. Include minimal code/examples ONLY if directly relevant to the context.
6. Maintain structured, well-formatted output.

---

Context:
{context}

Question:
{query}

---

### Output Format:
- **Answer:** Bullet points summarizing the requested information.
- **Explanation :** Step-by-step or structured explanation (flows, tables, or diagrams if suitable)Short, clear notes that explain the concept or workflow step‑by‑step and well formatted, suitable for an intermediate learner Explanation (Teacher Mode).
- **Code/Example (Optional):** Minimal illustrative snippet if relevant(only basic code, don't make it more complex. keep it as simple as it can be).
"""

    responce=model.invoke(prompt)
    answer=responce.content[0]['text']
    return answer,retrived_chunks
    
````
